# 🧪 LAB 3 — Age Classification from Video

## Objective
Detect faces in a video and estimate age.

## Implementation

```
import cv2
from deepface import DeepFace

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    try:
        analysis = DeepFace.analyze(frame, actions=['age'], enforce_detection=False)
        age = analysis[0]['age']
        cv2.putText(frame, f'Age: {age}', (50,50), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,255,0), 2)
    except:
        pass

    cv2.imshow('Age Detection', frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
