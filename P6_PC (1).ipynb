### Predicción de Demanda con el Modelo Holt-Winters

Ahora, aplicaremos un modelo de Suavizado Exponencial de Holt-Winters a nuestros datos de demanda. Este modelo es adecuado para series temporales que presentan tendencias y patrones estacionales. Entrenaremos el modelo con los datos de `train_data` y generaremos predicciones para el período de `test_data`.

```python
import numpy as np
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

# Establecer la columna 'FECHA' como índice para el modelado de series temporales
train_data_ts = train_data.set_index('FECHA')
test_data_ts = test_data.set_index('FECHA')

# Definir la variable objetivo
y_train = train_data_ts['DEMANDA REAL (Q)']
y_test = test_data_ts['DEMANDA REAL (Q)']

# Ajustar el modelo Holt-Winters
# Se asume una estacionalidad semanal (seasonal_periods=7) y componentes aditivos para la tendencia y la estacionalidad.
# 'initialization_method="estimated"' permite que el modelo estime los valores iniciales.
model_hw = ExponentialSmoothing(
    y_train,
    seasonal_periods=7,  # Cambiado a estacionalidad semanal (7 días)
    trend='add',         # Tendencia aditiva
    seasonal='add',      # Estacionalidad aditiva
    initialization_method="estimated"
).fit()

# Realizar predicciones en el conjunto de prueba
predictions_hw = model_hw.forecast(len(y_test))

# Evaluar el rendimiento del modelo
# Convertir a arrays de numpy para evitar problemas de alineación de índices al calcular métricas
mse_hw = mean_squared_error(y_test.values, predictions_hw.values)
rmse_hw = np.sqrt(mse_hw)
r2_hw = r2_score(y_test.values, predictions_hw.values)
mae_hw = mean_absolute_error(y_test.values, predictions_hw.values)
mape_hw = mean_absolute_percentage_error(y_test.values, predictions_hw.values)

# Imprimir los resultados de la evaluación
print("\n--- Evaluación del Modelo Holt-Winters ---")
print(f"Error Cuadrático Medio (MSE): {mse_hw:.2f}")
print(f"Coeficiente de Determinación (R²): {r2_hw:.2f}")
print(f"Raíz del Error Cuadrático Medio (RMSE): {rmse_hw:.2f}")
print(f"Error Absoluto Medio (MAE): {mae_hw:.2f}")
print(f"Error Porcentual Absoluto Medio (MAPE): {mape_hw:.2f}%")
```

### Visualización de Predicciones del Modelo Holt-Winters

Ahora, visualizaremos las predicciones del modelo Holt-Winters junto con los datos de entrenamiento y prueba para observar cómo se ajusta a la serie temporal.


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Graficar la base de datos original, los datos de entrenamiento, los datos de prueba y las predicciones del modelo Holt-Winters
plt.figure(figsize=(15, 7))
sns.lineplot(x=train_data['FECHA'], y=train_data['DEMANDA REAL (Q)'], label='Entrenamiento')
sns.lineplot(x=test_data['FECHA'], y=test_data['DEMANDA REAL (Q)'], label='Prueba')
sns.lineplot(x=test_data['FECHA'], y=predictions_hw, label='Predicciones Holt-Winters', color='green')
plt.title('Demanda Real y Predicciones del Modelo Holt-Winters')
plt.xlabel('Fecha')
plt.ylabel('Demanda Real (Q)')
plt.grid(True)
plt.xticks(rotation=45) # Rota las etiquetas del eje X para mejor legibilidad
plt.legend()
plt.tight_layout()
plt.show()

# Graficar las predicciones vs los valores reales solo para el conjunto de prueba
plt.figure(figsize=(15, 7))
sns.lineplot(x=test_data['FECHA'], y=test_data['DEMANDA REAL (Q)'], label='Valores Reales', color='blue')
sns.lineplot(x=test_data['FECHA'], y=predictions_hw, label='Predicciones Holt-Winters', color='red')
plt.title('Holt-Winters: Predicciones vs Valores Reales (Conjunto de Prueba)')
plt.xlabel('Fecha')
plt.ylabel('Demanda Real (Q)')
plt.grid(True)
plt.xticks(rotation=45) # Rota las etiquetas del eje X para mejor legibilidad
plt.legend()
plt.tight_layout()
plt.show()
```

### Predicción de Demanda con el Modelo LSTM

Ahora implementaremos un modelo de red neuronal de memoria a corto y largo plazo (LSTM) para predecir la demanda. Los modelos LSTM son muy efectivos para capturar patrones temporales complejos en series de tiempo.

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, r2_score
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

# 1. Preparar Datos para LSTM

# Extraer la serie de demanda de los conjuntos de entrenamiento y prueba
demand_train = train_data['DEMANDA REAL (Q)'].values.reshape(-1, 1)
demand_test = test_data['DEMANDA REAL (Q)'].values.reshape(-1, 1)

# Escalar los datos (ajustar el escalador solo con los datos de entrenamiento para evitar fuga de datos)
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_train = scaler.fit_transform(demand_train)
scaled_test = scaler.transform(demand_test)

# Función para crear secuencias para el modelo LSTM
def create_dataset(dataset, look_back=1):
    X, y = [], []
    for i in range(len(dataset) - look_back):
        a = dataset[i:(i + look_back), 0]
        X.append(a)
        y.append(dataset[i + look_back, 0])
    return np.array(X), np.array(y)

# Definir el período 'look_back' (cuántos pasos de tiempo anteriores se usarán para predecir el siguiente)
look_back = 7 # Usaremos los datos de los 7 días anteriores

# Crear conjuntos de datos de entrenamiento y prueba con secuencias
X_train_lstm, y_train_lstm = create_dataset(scaled_train, look_back)
X_test_lstm, y_test_lstm = create_dataset(scaled_test, look_back)

# Reformar la entrada para que sea [muestras, pasos de tiempo, características] para LSTM
X_train_lstm = np.reshape(X_train_lstm, (X_train_lstm.shape[0], X_train_lstm.shape[1], 1))
X_test_lstm = np.reshape(X_test_lstm, (X_test_lstm.shape[0], X_test_lstm.shape[1], 1))

print(f"Dimensiones de X_train_lstm: {X_train_lstm.shape}")
print(f"Dimensiones de y_train_lstm: {y_train_lstm.shape}")
print(f"Dimensiones de X_test_lstm: {X_test_lstm.shape}")
print(f"Dimensiones de y_test_lstm: {y_test_lstm.shape}")
```

```python
# 2. Construir el Modelo LSTM
model_lstm = Sequential()
model_lstm.add(LSTM(units=60, return_sequences=False, input_shape=(look_back, 1))) # 1 capa LSTM, 60 unidades
model_lstm.add(Dropout(0.2)) # Dropout de 0.2
model_lstm.add(Dense(1)) # Capa de salida para una única predicción

# Compilar el modelo con el optimizador Adam y una tasa de aprendizaje de 0.001
model_lstm.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001), loss='mean_squared_error')

# Mostrar resumen del modelo
model_lstm.summary()
```

```python
# 3. Entrenar el Modelo LSTM
print("Entrenando el modelo LSTM...")
history = model_lstm.fit(
    X_train_lstm, y_train_lstm,
    epochs=50, # Número de épocas: 50
    batch_size=32, # Tamaño del lote: 32
    verbose=1,
    shuffle=False # Importante para datos de series de tiempo
)
print("Entrenamiento del modelo LSTM completado.")
```

```python
# 4. Realizar Predicciones

# Realizar predicciones en el conjunto de entrenamiento y prueba
train_predict_scaled_lstm = model_lstm.predict(X_train_lstm)
test_predict_scaled_lstm = model_lstm.predict(X_test_lstm)

# Invertir la transformación de las predicciones a la escala original
train_predict_lstm = scaler.inverse_transform(train_predict_scaled_lstm)
test_predict_lstm = scaler.inverse_transform(test_predict_scaled_lstm)

# Invertir la transformación de los valores reales para la comparación
y_train_actual_lstm = scaler.inverse_transform(y_train_lstm.reshape(-1, 1))
y_test_actual_lstm = scaler.inverse_transform(y_test_lstm.reshape(-1, 1))
```

```python
# 5. Evaluar el Rendimiento del Modelo LSTM
mse_lstm = mean_squared_error(y_test_actual_lstm, test_predict_lstm)
rmse_lstm = np.sqrt(mse_lstm)
r2_lstm = r2_score(y_test_actual_lstm, test_predict_lstm)
mae_lstm = mean_absolute_error(y_test_actual_lstm, test_predict_lstm)
mape_lstm = mean_absolute_percentage_error(y_test_actual_lstm, test_predict_lstm)

print("\n--- Evaluación del Modelo LSTM ---")
print(f"Error Cuadrático Medio (MSE): {mse_lstm:.2f}")
print(f"Raíz del Error Cuadrático Medio (RMSE): {rmse_lstm:.2f}")
print(f"Coeficiente de Determinación (R²): {r2_lstm:.2f}")
print(f"Error Absoluto Medio (MAE): {mae_lstm:.2f}")
print(f"Error Porcentual Absoluto Medio (MAPE): {mape_lstm:.2f}%")
```

### Predicción de Demanda con un Modelo SimpleRNN

Una Red Neuronal Recurrente (RNN) es un tipo de red neuronal diseñada para reconocer patrones en secuencias de datos, como series temporales. A diferencia de las redes neuronales feed-forward tradicionales, las RNN tienen "memoria" incorporada, lo que les permite utilizar información de pasos de tiempo anteriores para influir en la salida del paso de tiempo actual.

Aquí implementaremos un modelo `SimpleRNN` de Keras, que es una versión básica de una RNN.

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, r2_score
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense, Dropout
import matplotlib.pyplot as plt
import seaborn as sns

# Los datos ya están preparados por las celdas anteriores (scaled_train, scaled_test, X_train_lstm, y_train_lstm, X_test_lstm, y_test_lstm, look_back, scaler).
# Si estas variables no estuvieran disponibles, sería necesario ejecutar las celdas de preparación de datos para LSTM primero.

# 1. Construir el Modelo SimpleRNN
model_rnn = Sequential()
model_rnn.add(SimpleRNN(units=60, return_sequences=False, input_shape=(look_back, 1))) # 1 capa SimpleRNN, 60 unidades
model_rnn.add(Dropout(0.2)) # Dropout de 0.2
model_rnn.add(Dense(1)) # Capa de salida para una única predicción

# Compilar el modelo con el optimizador Adam y una tasa de aprendizaje de 0.001
model_rnn.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001), loss='mean_squared_error')

# Mostrar resumen del modelo
model_rnn.summary()
```

```python
# 2. Entrenar el Modelo SimpleRNN
print("Entrenando el modelo SimpleRNN...")
history_rnn = model_rnn.fit(
    X_train_lstm, y_train_lstm,
    epochs=50, # Usar el mismo número de épocas que LSTM para comparación
    batch_size=32, # Usar el mismo tamaño de lote
    verbose=1,
    shuffle=False # Importante para datos de series de tiempo
)
print("Entrenamiento del modelo SimpleRNN completado.")
```

```python
# 3. Realizar Predicciones y Evaluar el Modelo

# Realizar predicciones en el conjunto de entrenamiento y prueba
train_predict_scaled_rnn = model_rnn.predict(X_train_lstm)
test_predict_scaled_rnn = model_rnn.predict(X_test_lstm)

# Invertir la transformación de las predicciones a la escala original
train_predict_rnn = scaler.inverse_transform(train_predict_scaled_rnn)
test_predict_rnn = scaler.inverse_transform(test_predict_scaled_rnn)

# Los valores reales y_train_actual_lstm y y_test_actual_lstm ya están disponibles de la sección LSTM.

# Evaluar el rendimiento del modelo SimpleRNN
mse_rnn = mean_squared_error(y_test_actual_lstm, test_predict_rnn)
rmse_rnn = np.sqrt(mse_rnn)
r2_rnn = r2_score(y_test_actual_lstm, test_predict_rnn)
mae_rnn = mean_absolute_error(y_test_actual_lstm, test_predict_rnn)
mape_rnn = mean_absolute_percentage_error(y_test_actual_lstm, test_predict_rnn)

print("\n--- Evaluación del Modelo SimpleRNN ---")
print(f"Error Cuadrático Medio (MSE): {mse_rnn:.2f}")
print(f"Raíz del Error Cuadrático Medio (RMSE): {rmse_rnn:.2f}")
print(f"Coeficiente de Determinación (R²): {r2_rnn:.2f}")
print(f"Error Absoluto Medio (MAE): {mae_rnn:.2f}")
print(f"Error Porcentual Absoluto Medio (MAPE): {mape_rnn:.2f}%")
```

### Predicción de Demanda con el Modelo XGBoost

XGBoost (eXtreme Gradient Boosting) es una implementación optimizada de árboles de decisión basada en el algoritmo de *gradient boosting*. Es conocido por su velocidad y rendimiento, siendo una opción popular para problemas de regresión y clasificación en datos tabulares. Aquí, lo aplicaremos a nuestra serie temporal de demanda.

```python
import xgboost as xgb
import numpy as np
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
import seaborn as sns

# Preparar los datos para XGBoost
# X serán las características (FECHA_NUM) e y será la variable objetivo (DEMANDA REAL (Q))
X_train_xgb = train_data[['FECHA_NUM']]
y_train_xgb = train_data['DEMANDA REAL (Q)']
X_test_xgb = test_data[['FECHA_NUM']]
y_test_xgb = test_data['DEMANDA REAL (Q)']

# 1. Construir y Entrenar el Modelo XGBoost
# Se utiliza xgb.XGBRegressor para problemas de regresión
model_xgb = xgb.XGBRegressor(
    objective='reg:squarederror', # Función objetivo para regresión (minimiza el error cuadrático)
    n_estimators=100,             # Número de árboles en el modelo
    learning_rate=0.1,            # Tasa de aprendizaje
    max_depth=5,                  # Profundidad máxima de cada árbol
    random_state=42               # Para reproducibilidad
)

print("Entrenando el modelo XGBoost...")
model_xgb.fit(X_train_xgb, y_train_xgb)
print("Entrenamiento del modelo XGBoost completado.")
```

```python
# 2. Realizar Predicciones
predictions_xgb = model_xgb.predict(X_test_xgb)

# 3. Evaluar el Rendimiento del Modelo XGBoost
mse_xgb = mean_squared_error(y_test_xgb, predictions_xgb)
rmse_xgb = np.sqrt(mse_xgb)
r2_xgb = r2_score(y_test_xgb, predictions_xgb)
mae_xgb = mean_absolute_error(y_test_xgb, predictions_xgb)
mape_xgb = mean_absolute_percentage_error(y_test_xgb, predictions_xgb)

print("\n--- Evaluación del Modelo XGBoost ---")
print(f"Error Cuadrático Medio (MSE): {mse_xgb:.2f}")
print(f"Raíz del Error Cuadrático Medio (RMSE): {rmse_xgb:.2f}")
print(f"Coeficiente de Determinación (R²): {r2_xgb:.2f}")
print(f"Error Absoluto Medio (MAE): {mae_xgb:.2f}")
print(f"Error Porcentual Absoluto Medio (MAPE): {mape_xgb:.2f}%")
```

### Visualización de Predicciones del Modelo XGBoost

```python
# Graficar la Demanda Real y las Predicciones del Modelo XGBoost (conjunto de prueba)
plt.figure(figsize=(15, 7))
sns.lineplot(x=test_data['FECHA'], y=y_test_xgb, label='Valores Reales (Prueba)', color='blue')
sns.lineplot(x=test_data['FECHA'], y=predictions_xgb, label='Predicciones XGBoost (Prueba)', color='red')
plt.title('XGBoost: Predicciones vs Valores Reales (Conjunto de Prueba)')
plt.xlabel('Fecha')
plt.ylabel('Demanda Real (Q)')
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# **Verificación de seasonal_periods para el Modelo Holt-Winters**

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Agrupar la demanda por fecha para obtener la demanda diaria total
daily_demand = df_cleaned.groupby('FECHA')['DEMANDA REAL (Q)'].sum().reset_index()

# Extraer el día de la semana y el mes
daily_demand['DAY_OF_WEEK'] = daily_demand['FECHA'].dt.dayofweek # Lunes=0, Domingo=6
daily_demand['MONTH'] = daily_demand['FECHA'].dt.month

plt.figure(figsize=(16, 6))

# Gráfico de demanda promedio por día de la semana
plt.subplot(1, 2, 1)
sns.lineplot(x='DAY_OF_WEEK', y='DEMANDA REAL (Q)', data=daily_demand.groupby('DAY_OF_WEEK')['DEMANDA REAL (Q)'].mean().reset_index())
plt.title('Demanda Promedio por Día de la Semana')
plt.xlabel('Día de la Semana (0=Lunes, 6=Domingo)')
plt.ylabel('Demanda Real (Q) Promedio')
plt.xticks(ticks=range(7), labels=['Lun', 'Mar', 'Mié', 'Jue', 'Vie', 'Sáb', 'Dom'])
plt.grid(True)

# Gráfico de demanda promedio mensual
plt.subplot(1, 2, 2)
sns.lineplot(x='MONTH', y='DEMANDA REAL (Q)', data=daily_demand.groupby('MONTH')['DEMANDA REAL (Q)'].mean().reset_index())
plt.title('Demanda Promedio Mensual')
plt.xlabel('Mes')
plt.ylabel('Demanda Real (Q) Promedio')
plt.grid(True)
plt.tight_layout()
plt.show()
```

El valor de `seasonal_periods=2` que se usó anteriormente es un valor arbitrario y no refleja una estacionalidad típica. Una vez que hayas revisado los gráficos, podríamos ajustar el `seasonal_periods` a 7, o a otro valor que los gráficos sugieran fuertemente.
